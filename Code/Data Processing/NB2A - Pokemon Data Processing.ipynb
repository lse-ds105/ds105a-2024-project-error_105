```python
range_of_poke_id = range(gen_1_poke_df['pokemon_id'].min(), gen_1_poke_df['pokemon_id'].max()+1)
```

```python
def extract_pokemon_habitat(): 
    json_data = {
    pokemon_id: json.load(open(f'../../data/Pokemon Data/generation_1_species_details/pokemon_{pokemon_id}_species_details', 'r'))
    for pokemon_id in range_of_poke_id
    }
    gen_1_poke_df['habitat_name'] = gen_1_poke_df['pokemon_id'].apply(lambda x: json_data[x]['habitat']['name'])

extract_pokemon_habitat()
```

```python
def extract_pokemon_stats_and_types():
    json_data = {
    pokemon_id: json.load(open(f'../../data/Pokemon Data/generation_1_pokemon_details/pokemon_{pokemon_id}_details', 'r'))
    for pokemon_id in range_of_poke_id
    }
    gen_1_poke_df['type_1'] = gen_1_poke_df['pokemon_id'].apply(lambda x: json_data[x]['types'][0]['type']['name'])
    gen_1_poke_df['type_2'] = gen_1_poke_df['pokemon_id'].apply(lambda x: json_data[x]['types'][1]['type']['name'] if len(json_data[x]['types']) > 1 else None)
    gen_1_poke_df['hp_stat'] = gen_1_poke_df['pokemon_id'].apply(lambda x: json_data[x]['stats'][0]['base_stat'])
    gen_1_poke_df['attack_stat'] = gen_1_poke_df['pokemon_id'].apply(lambda x: json_data[x]['stats'][1]['base_stat'])
    gen_1_poke_df['defense_stat'] = gen_1_poke_df['pokemon_id'].apply(lambda x: json_data[x]['stats'][2]['base_stat'])
    gen_1_poke_df['special_attack_stat'] = gen_1_poke_df['pokemon_id'].apply(lambda x: json_data[x]['stats'][3]['base_stat'])
    gen_1_poke_df['special_defence_stat'] = gen_1_poke_df['pokemon_id'].apply(lambda x: json_data[x]['stats'][4]['base_stat'])
    gen_1_poke_df['speed_stat'] = gen_1_poke_df['pokemon_id'].apply(lambda x: json_data[x]['stats'][5]['base_stat'])
    return 
    
extract_pokemon_stats_and_types()
```
